```python
import numpy as np
import tensorflow as tf
import os

# Set environment variable to use legacy Keras if TFRS requires it
os.environ["TF_USE_LEGACY_KERAS"] = "1"
print(f"TF_USE_LEGACY_KERAS environment variable set to: {os.environ.get('TF_USE_LEGACY_KERAS')}")

# If you still encounter an ImportError regarding Keras version after running this cell,
# please restart the Colab runtime (Runtime -> Restart runtime) from the menu.
# This ensures the environment variable takes full effect for all imports.

# Install tensorflow-recommenders if it's not already installed and then import
try:
    import tensorflow_recommenders as tfrs
except ImportError:
    !pip install -qq tensorflow-recommenders
    import tensorflow_recommenders as tfrs

# 1. Prepare dummy data (Users and the Movies they watched)
data = {
    "user_id": np.array(["user_1", "user_2", "user_3", "user_1"]),
    "movie_title": np.array(["Toy Story", "The Matrix", "Toy Story", "The Matrix"])
}

# Convert to TensorFlow datasets
users_dataset = tf.data.Dataset.from_tensor_slices(data["user_id"])
movies_dataset = tf.data.Dataset.from_tensor_slices(data["movie_title"])

# Pair them together for training
ratings_dataset = tf.data.Dataset.from_tensor_slices((data["user_id"], data["movie_title"]))
ratings_dataset = ratings_dataset.map(lambda user, movie: {"user_id": user, "movie_title": movie})

# 2. Define the Unique Vocabulary for Embeddings
unique_user_ids = np.unique(data["user_id"])
unique_movie_titles = np.unique(data["movie_title"])

# 3. Build the Two-Tower Model
class MovieModel(tfrs.Model):
    def __init__(self):
        super().__init__()

        # --- User Tower ---
        self.user_model = tf.keras.Sequential([
            tf.keras.layers.StringLookup(vocabulary=unique_user_ids, mask_token=None),
            tf.keras.layers.Embedding(len(unique_user_ids) + 1, 32) # Map to 32-dim vector
        ])

        # --- Item (Movie) Tower ---
        self.movie_model = tf.keras.Sequential([
            tf.keras.layers.StringLookup(vocabulary=unique_movie_titles, mask_token=None),
            tf.keras.layers.Embedding(len(unique_movie_titles) + 1, 32) # Map to 32-dim vector
        ])

        # --- Task Layer (Computes mathematical similarity) ---
        metrics = tfrs.metrics.FactorizedTopK(candidates=movies_dataset.batch(128).map(self.movie_model))
        self.task = tfrs.tasks.Retrieval(metrics=metrics)

    def compute_loss(self, features, training=False):
        # Pass data through the towers to get embeddings
        user_embeddings = self.user_model(features["user_id"])
        movie_embeddings = self.movie_model(features["movie_title"])

        # Calculate how close they are
        return self.task(user_embeddings, movie_embeddings)

# 4. Train the Model
model = MovieModel()
model.compile(optimizer=tf.keras.optimizers.Adagrad(learning_rate=0.1))
model.fit(ratings_dataset.batch(4096), epochs=5, verbose=0)

# 5. Setup Brute-Force Search for Recommendations
index = tfrs.layers.factorized_top_k.BruteForce(model.user_model)
index.index_from_dataset(movies_dataset.batch(100).map(lambda title: (title, model.movie_model(title))))

# Test the system for "user_2"
# Explicitly set k=1 to get the top recommendation, as there are only 4 candidates and default k is often 10.
_, titles = index(np.array(["user_2"]), k=1)
print(f"Top recommendation for User 2: {titles[0, 0].numpy().decode('utf-8')}")
```
